### Question 1: Import TensorFlow and Display Its Version

```python
import tensorflow as tf

print("TensorFlow Version:", tf.__version__)
```

### Question 2: Create Scalar, Vector, and Matrix Tensors

```python
scalar = tf.constant(10)
vector = tf.constant([10, 20, 30])
matrix = tf.constant([[1, 2], [3, 4]])
```

### Question 3: Display the Value, Shape, and Data Type of Each Tensor

```python
print("Scalar:", scalar.numpy())
print("Shape:", scalar.shape)
print("Data Type:", scalar.dtype)

print("\nVector:", vector.numpy())
print("Shape:", vector.shape)
print("Data Type:", vector.dtype)

print("\nMatrix:\n", matrix.numpy())
print("Shape:", matrix.shape)
print("Data Type:", matrix.dtype)
```

### Question 4: Create Two Matrices and Perform Addition, Subtraction, Element-wise Multiplication, and Matrix Multiplication

```python
A = tf.constant([[1, 2], [3, 4]])
B = tf.constant([[5, 6], [7, 8]])

print("Addition:\n", tf.add(A, B).numpy())
print("Subtraction:\n", tf.subtract(A, B).numpy())
print("Element-wise Multiplication:\n", tf.multiply(A, B).numpy())
print("Matrix Multiplication:\n", tf.matmul(A, B).numpy())
```

### Question 5: Observe the Difference Between Element-wise Multiplication and Matrix Multiplication

```python
print("Element-wise Multiplication:\n", tf.multiply(A, B).numpy())
print("Matrix Multiplication:\n", tf.matmul(A, B).numpy())
```

**Observation:**

- **Element-wise multiplication:** Multiplies corresponding elements of two matrices.
- **Matrix multiplication:** Multiplies rows of the first matrix with columns of the second matrix and adds their products.

Both operations give different results.

### Question 2: Create a Simple Sequential Model

```python
model = keras.Sequential()
```

### Question 3: Add the Following Layers
- An input layer with **2 input features**
- A hidden layer with **4 neurons using ReLU activation**
- An output layer with **1 neuron using Sigmoid activation**

```python
model.add(keras.layers.Input(shape=(2,)))
model.add(keras.layers.Dense(4, activation='relu'))
model.add(keras.layers.Dense(1, activation='sigmoid'))
```
